## Applying functions to columns and creating new columns

We need to create another column in d3, which contains  a boolean value for each city to indicate whether it's a union territory or not. 
- HINT: Chandigarh, Pondicherry and Delhi are only 3 union territories here. 


```python

# write function here
def is_ut(x):
    return x in ['Chandigarh', 'Delhi', 'Pondicherry']

# apply function
d3['is_ut'] = d3['city'].apply(is_ut)

print(d3)
```

HINT - 
- Define a function c_to_f which takes input temp in celsius and returns a value with temperature in Fahrenheit.
- To check: c_to_f(10) should return 50. 


```python
# write function here

# define function
def c_to_f(temp):
    return (temp * 9/5) + 32

# check
print(c_to_f(10))   


```

## Groupby

```python
# Create a dataframe using dictionary of your choice

data = {
    'name': ['Arooba', 'Ali', 'Sara', 'John'],
    'age': [24, 22, 23, 25],
    'marks': [85, 90, 88, 92]
}

df_new = pd.DataFrame(data)
print(df_new)
```

```python
# Use Groupby of single column with aggregate sum()
grouped = df_new.groupby('name').sum()
print(grouped)

```

```python
# Use Groupby of single column with aggregate count()
grouped = df_new.groupby('name').count()
print(grouped)
```

```python
# Use Groupby of single column with aggregate min() and max()


grouped_min = df_new.groupby('name').min()
grouped_max = df_new.groupby('name').max()

print(grouped_min)
print(grouped_max)
```

```python
# Use Groupby of any 2 columns with aggregate mean()
df_new = pd.DataFrame({
    'category': ['A', 'B', 'A', 'B'],
    'marks': [85, 90, 88, 92]
})

grouped = df_new.groupby(['category']).mean()
print(grouped)
```

```python
# Use Groupby of any 2 columns with aggregate min() and max()

import pandas as pd

data = {
    'city': ['Delhi', 'Delhi', 'Mumbai', 'Mumbai', 'Pune'],
    'type': ['A', 'B', 'A', 'B', 'A'],
    'marks': [85, 90, 88, 92, 75]
}

df_new = pd.DataFrame(data)

grouped = df_new.groupby(['city', 'type']).agg(['min', 'max'])
print(grouped)
```
